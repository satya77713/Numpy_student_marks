```python
import numpy as np
```

```python
student_marks = np.array([85, 90, 78, 92, 88, 76, 95, 89, 84, 91])

print("Student Marks:", student_marks)
```

```python
mean_marks = np.mean(student_marks)
print(f"mean marks : {mean_marks}")
```

```python
median_marks = np.median(student_marks)
print(f"median marks : {median_marks}")
```

```python
minimum_marks = np.min(student_marks)
print(f"minimum marks : {minimum_marks}")
```

```python
maximum_marks = np.max(student_marks)
print(f"maximum marks : {maximum_marks}")
```

```python
standard_deviation = np.std(student_marks)
print(f"standard deviation : {standard_deviation}")
```

```python
print(f"mean marks : {mean_marks}")
print(f"median marks : {median_marks}")
print(f"minimum marks : {minimum_marks}")
print(f"maximum marks : {maximum_marks}")
print(f"standard deviation : {standard_deviation}")
```
